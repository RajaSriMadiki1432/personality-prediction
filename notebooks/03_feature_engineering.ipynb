## Load Cleaned Data

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/processed/cleaned_personality_data.csv')
trait_scores = ['EXT_score','EST_score','AGR_score','CSN_score','OPN_score']
personality_columns = [f'{t}{i}' for t in ['EXT','EST','AGR','CSN','OPN'] for i in range(1,11)]
```

## Features for clustering

```python
X_cluster = df[trait_scores].copy()
```

## Scale for clustering

```python

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_cluster_scaled = scaler.fit_transform(X_cluster)
```

```python
print(df[trait_scores].corr())
```

```python
short_form_items = {
    'EXT': ['EXT1','EXT2','EXT3','EXT4'],
    'EST': ['EST1','EST2','EST3','EST4'],
    'AGR': ['AGR1','AGR2','AGR3','AGR4'],
    'CSN': ['CSN1','CSN2','CSN3','CSN4'],
    'OPN': ['OPN1','OPN2','OPN3','OPN4'],
}

X_short = df[[item for items in short_form_items.values() for item in items]]
y_short = df[trait_scores]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X_short,y_short,test_size=0.2,random_state=42)
```

```python
##save engineered features

import joblib
X_cluster_scaled_df = pd.DataFrame(X_cluster_scaled, columns=trait_scores)
X_cluster_scaled_df.to_csv('../data/processed/cluster_features.csv',index=False)

X_train.to_csv('../data/processed/X_train.csv',index=False)
X_test.to_csv('../data/processed/X_test.csv', index=False)
y_train.to_csv('../data/processed/y_train.csv', index=False)
y_test.to_csv('../data/processed/y_test.csv', index=False)

joblib.dump(scaler, '../models/cluster_scaler.pkl')
```
