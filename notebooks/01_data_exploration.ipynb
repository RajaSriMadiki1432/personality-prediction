# Step 9 — Check unique values

```python
personality_columns= [
    col for col in df.columns
    if col.startswith(("EXT","EST","AGR","CSN","OPN"))
    and not col.endswith("_E")
]

print("Number of personalty question columns:",len(personality_columns))
print(personality_columns)
```

```python
#unique values in the personality responses 
for col in personality_columns:
    print(col,sorted(df[col].dropna().unique()))
```

# Step-10  Check the five personality groups

```python
groups = {
    "Extraversion": [f"EXT{i}" for i in range(1,11)],
    "Neuroticism": [f"EST{i}" for i in range(1,11)],
    "Agreeableness": [f"AGR{i}" for i in range(1,11)],
    "Conscientiousness": [f"CSN{i}" for i in range(1,11)],
    "Openness": [f"OPN{i}" for i in range(1,11)]
}

for trait, columns in groups.items():
    print(trait, ":", len(columns), "questions")
```

# Step-14  Find the overall average for each personality trait

```python
#calculate average score for each big five trait

for trait, columns in groups.items():
    df[trait + "_score"]= df[columns].mean(axis=1)

print(df[
    [
        "Extraversion_score",
        "Neuroticism_score",
        "Agreeableness_score",
        "Conscientiousness_score",
        "Openness_score"
    ]
].head())
```

# Step-15  Visualize the five tarits

```python
import matplotlib.pyplot as plt

trait_columns = [
     "Extraversion_score",
    "Neuroticism_score",
    "Agreeableness_score",
    "Conscientiousness_score",
    "Openness_score"
]

df[trait_columns].boxplot(figsize=(10,6))

plt.title("Distribution of Big Five Personality Scores")
plt.ylabel("Average Response")
plt.xticks(rotation=20)
plt.show()
```

# Step 16: Analyze the Big Five Personality Scores

```python
# Calculate the average score for each personality trait

trait_means= df[trait_columns].mean()

print("Average score for each personality trait:\n")
print(trait_means)
```

# Step 17: Visualize Average Personality Scores

```python
# plot the average score of each personality trait

trait_means.plot(kind="bar",figsize=(10,6))

plt.title("Average Big Five Personality Scores")
plt.xlabel("Personality Trait")
plt.ylabel("Average Score")
plt.show()
```

# Step 18: Check Correlation Between Personality Traits

```python
# calculate correlation between personality traits

correlation = df[trait_columns].corr()

print(correlation)
```

```python
#visualizing

plt.figure(figsize=(10,6))

sns.heatmap(correlation,annot=True,fmt=".2f")

plt.title("Correlation Between Big Five Personality Traits")
plt.show()
```

# Step 19: Check the Distribution of Each Trait

```python
# Create Big Five personality trait scores

df["Extraversion"] = df[
    ["EXT1", "EXT2", "EXT3", "EXT4", "EXT5",
     "EXT6", "EXT7", "EXT8", "EXT9", "EXT10"]
].mean(axis=1)

df["Agreeableness"] = df[
    ["AGR1", "AGR2", "AGR3", "AGR4", "AGR5",
     "AGR6", "AGR7", "AGR8", "AGR9", "AGR10"]
].mean(axis=1)

df["Conscientiousness"] = df[
    ["CSN1", "CSN2", "CSN3", "CSN4", "CSN5",
     "CSN6", "CSN7", "CSN8", "CSN9", "CSN10"]
].mean(axis=1)

df["Neuroticism"] = df[
    ["EST1", "EST2", "EST3", "EST4", "EST5",
     "EST6", "EST7", "EST8", "EST9", "EST10"]
].mean(axis=1)

df["Openness"] = df[
    ["OPN1", "OPN2", "OPN3", "OPN4", "OPN5",
     "OPN6", "OPN7", "OPN8", "OPN9", "OPN10"]
].mean(axis=1)

print("Big Five scores created successfully!")
print(df[[
    "Extraversion",
    "Agreeableness",
    "Conscientiousness",
    "Neuroticism",
    "Openness"
]].head())
```
